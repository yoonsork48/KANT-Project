```python
# 1. 상품 클래스
class Product:
    # 1-1. 상품 번호 자동 부여
    next_number = 1
    # 1-2. 속성
    def __init__(self, name, price, qty):
    # 1-2-1. 제품 넘버 부여
        self.number = Product.next_number
        Product.next_number += 1
    # 1-2-2. 제품명
        self.name = name
    # 1-2-3. 가격
        self.price = price
    # 1-2-4. 재고
        self.qty = qty

# # 쇼핑몰 생성
# mall = ShoppingCart()
# # 상품 5개 생성
# product_1 = Product("무선 이어폰", 89000, 3)
# product_2 = Product("텀블러", 15000, 10)
# product_3 = Product("보조배터리", 32000, 5)
# product_4 = Product("마우스패드", 9000, 0)
# product_5 = Product("usb 허브", 18000, 4)

# 2. 장바구니 클래스
class ShoppingCart:
    def __init__(self):     #장바구니 생성
        self.bucket = {}    #장바구니 비우기

    # 장바구니에 담기
    def add(self, number, qty, products):
        #상품조회
        product = products.get(number)

        if product is None:     # 엉뚱한 숫자 클릭
            print("존재하지 않는 상품입니다.")
            return 

        if product.qty == 0:    # 눌렀는데 품절
            print("품절입니다.")
            return

        if product.qty <= 0:    # 0 이거나 음수 클릭
            print("수량은 1개 이상이어야 합니다.")
            return
        
        # 이미 담긴 수량(카트에 담긴 양)
        current = self.bucket.get(number, 0)

        # 재고 = 상품 총량 - 카트에 담긴 양
        available = product.qty - current

        # available = product.qty - self.bucket.get(product.number, 0)

        if qty > available:     # 구매 물량이 재고보다 많을때
            print(f"재고가 부족합니다. 가능 수량 : {available}개")
            return

        # 장바구니에 수량 추가
        self.bucket[number] = current + qty
        print(f"{product.name}{qty}개를 담았습니다.")
        
    # 장바구니에서 빼기
    def remove(self, number, qty, products):
        if number not in self.bucket:   #엉뚱한 숫자 클릭
            print("장바구니에 없는 상품입니다.")
            return
        if qty <= 0:    # 0이나 음수 클릭
            print("상품은 1개 이상이어야 합니다.")
            return

        # 카트안에 든 양 = 즉, 카트 생성
        current = self.bucket[number]

        if qty > current:   # 뺄 물량이 카트에 든 양보다 많을때
            print(f"담긴 수량이 부족합니다. 현재 {current}개")
            return

        # 이미 물량이 들어있는 카트에서 차감을 시키겠다.
        self.bucket[number] -= qty

        # 수량이 0이면 상품 삭제
        if self.bucket[number] == 0:
            del self.bucket[number]

        print(f"{products[number].name}{qty}개를 뺐습니다.")

    # 상품의 합계 계산
    def subtotal(self, products):
        total = 0

        for number, qty in self.bucket.items():
            product = products[number]
            total += product.price * qty

        return total

    # 영수증 및 장바구니 출력
    def show (self, products):

        if not self.bucket:
            print("장바구니가 비어있습니다.")
            return

        print("\n" + "-" * 45)

        for number, qty in self.bucket.items():
            product = products[number]
            price = product.price * qty

            print(f"{product.nbame:<12} x {qty:<3} {price>10,}원")

        print("-" * 45)

        # 상품 합계
        total = self.subtotal(products)

class ShoppingMall:
    def __init__(self):
        self.product_list = [
            Product('무선 이어폰', 89000, 3),
            Product('텀블러', 15000, 10),
            Product('보조배터리', 32000, 5),
            Product('마우스패드', 9000, 0),
            Product('USB 허브', 18000, 4),
        ]
    
    #상품 번호를 기준으로 조회 되도록 저장
    self.products = {}

    for product in self.product_list:
        self.proudcts[product.number] = product

        # 장바구니 객체 생성
        self.cart = ShoppingCart()

    # 상품 목록 출력
    def show_products(self):

        print("\n<<<<<<<<<<<<<<< 상품 목록 >>>>>>>>>>>>>>>>>")

        for number, product in self.products.items():     #숫자로 상품 보여주기

            if product.qty == 0:
                stock = "품절"
            else:
                stock = f"재고 {product.qty}개"

            print(
                f"{product.number}. {product.name:<12} "
                f"{product.price:>8,}원 ({stock})"
            )

    # 숫자 입력
    def get_number(self, message):

        try:
            return int(input(message))
        except ValueError:
            print("숫자를 제대로 입력해주세요.")
            return None

# # 쿠폰 클래스
# class Coupon:
#     def __init__(self): # 쿠폰 번호별 네이밍
#         print("1. 쿠폰 없음")
#         print("2. 10% 할인(최대 5,000원)")
#         print("3. 3,000원 할인(20,000원 이상 구매시)")

#         number = input("쿠폰 번호: ") # 쿠폰 번호 입력

#         if number in ("1", "2", "3"): # 입력 가능한 쿠폰번호
#             self.number = int(number)

#             if self.number == 1: # 1번쿠폰 적용
#                 print("적용할 쿠폰이 없습니다.")
#             else: # 2 ,3번 쿠폰 적용시 출력
#                 print(f"[{self.get_name()}]쿠폰을 적용했습니다.")
#         else: # 쿠폰 오류
#             print("잘못된 쿠폰 번호 입니다.")

#     def get_name(self):
#         if self.number == 2: # 2번쿠폰 적용
#             return __init__(self, rate=0.1, max_amount=5000)
#             print("(10% 할인(최대 5,000원))")
#             # self.name = f"{rate:.0%}할인 (최대{max_amount:,}원)"
            
#         elif self.number == 3: # 3번쿠폰 적용
#             return __init__(self, amount=3000, min_total=20000)
#             print("3,000원 할인(20,000원 이상 구매시)") 
#             # self.name = f"{amount:,}원 할인 ({min_amount:,}원 이상 구매시)"
#         return "쿠폰없음" #쿠폰 적용 안함(없음)처리
    

# ============쇼핑몰 실행=================
    def run(self):
    
        while True:
            # 첫 실행화면
            print("\n" + "-" * 56)
            print("1.상품목록 2.담기 3.빼기 4.장바구니 5.쿠폰 6.결제 q.종료")
            print("-" * 56)

            user_input = input("선택: ").strip().lower()

            # 1. 상품목록
            if user_input == "1":
                print(f"무선 이어폰     89,000원  (재고:  3개)" )
                print(f"텀블러          15,000원  (재고: 10개)" )
                print(f"보조배터리      32,000원  (재고:  5개)" )
                print(f"마우스패드       9,000원  (---품절---)" )
                print(f"USB 허브        18,000원  (재고:  4개)" )

            # 2. 상품 담기
            elif user_input == "2":
                number = self.get_number("상품 번호: ")

                if number is None:
                    print("존재하지 않는 상품입니다.")
                    continue

                Product = self.products.get(number)

                if Product is None:
                    print("존재하지 않는 상품입니다.")
                    continue

                if Product.qty == 0:
                    print("품절된 상품입니다.")
                    continue

                qty = self.get_number("수량: ")

                if qty is not None:
                   self.bucket.add(number, qty, self.products)

            # 3. 상품 빼기
            elif user_input == "3":
                number = self.get_numnber("상품번호: ")

                if number is None:
                    continue

                qty = self.get_number("뺄 수량: ")

                if qty is not None:
                   self.bucket.add(number, qty, self.products)

            # 4. 장바구니
            elif user_input == "4":
                self.cart.show(self.products, self.coupon)
            
             # 6. 결제
            elif user_input == "6":
                self.cart.checkout(self.products, self.coupon)

            # 종료
            elif user_input == "q":
                print("쇼핑몰을 종료합니다.")
                break

            else:
                print("다시 선택해 주세요.")

# ===================프로그램 실행======================
mall = ShoppingMall()
mall.run()
```

```python
class Product:
    """판매 상품의 정보와 실제 재고를 관리한다."""
    next_number = 1

    def __init__(self, name, price, qty):
        if price < 0 or qty < 0:
            raise ValueError("가격과 재고는 음수일 수 없습니다.")

        self.number = Product.next_number
        Product.next_number += 1
        self.name = name
        self.price = price
        self.__qty = qty

    @property
    def qty(self):
        # 외부에서는 조회만 허용한다.
        return self.__qty

    def decrease_stock(self, qty):
        # 유효한 수량이고 재고가 충분할 때만 변경한다.
        if qty <= 0 or qty > self.__qty:
            return False
        self.__qty -= qty
        return True


class ShoppingCart:
    """{상품 번호: 담은 수량} 형태로 장바구니를 관리한다."""
    def __init__(self):
        self.bucket = {}

    def add(self, number, qty, products):
        product = products.get(number)
        if product is None:
            print("존재하지 않는 상품입니다.")
            return
        if qty <= 0:
            print("수량은 1개 이상이어야 합니다.")
            return
        if product.qty == 0:
            print("품절입니다.")
            return

        current = self.bucket.get(number, 0)
        available = product.qty - current
        if qty > available:
            print(f"재고가 부족합니다. 추가 가능 수량: {available}개")
            return

        self.bucket[number] = current + qty
        print(f"{product.name} {qty}개를 담았습니다.")

    def remove(self, number, qty, products):
        if number not in self.bucket:
            print("장바구니에 없는 상품입니다.")
            return
        if qty <= 0:
            print("수량은 1개 이상이어야 합니다.")
            return

        current = self.bucket[number]
        if qty > current:
            print(f"담긴 수량이 부족합니다. 현재 {current}개")
            return

        self.bucket[number] -= qty
        if self.bucket[number] == 0:
            del self.bucket[number]
        print(f"{products[number].name} {qty}개를 뺐습니다.")

    def subtotal(self, products):
        total = 0
        for number, qty in self.bucket.items():
            product = products[number]
            total += product.price * qty
        return total

    def show(self, products):
        if not self.bucket:
            print("장바구니가 비어 있습니다.")
            return

        print("\n" + "-" * 45)
        print("[장바구니]")
        for number, qty in self.bucket.items():
            product = products[number]
            price = product.price * qty
            print(f"{number}. {product.name:<12} x {qty:<3} {price:>10,}원")
        print("-" * 45)
        print(f"총 상품 금액: {self.subtotal(products):,}원")

    def checkout(self, products):
        if not self.bucket:
            print("장바구니가 비어 있어 결제할 수 없습니다.")
            return False
```
