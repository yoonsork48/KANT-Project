```python
# 부모 클래스 Base Class , Super Class , Parent Class
class Animal:
    # 공통 기능
    def breathe(self):
        print("숨을 쉽니다.")
    def sound(self):
        print("소리를 냅니다.")

# 상속 (Inheritance)
class Dog(Animal):
    # 상속받은 클래스에서만 있는 고유 기능
    def bark(self):
        print("멍멍")

# 인스턴스 생성
dog1 = Dog()
dog1.breathe()  # 부모 클래스로 부터 물려받은 메서드를 그대로 사용할 수 있다.
dog1.bark()    # 자식 클래스의 고유기능

# 메서드 오버라이딩 (method Overriding)
# 부모 클래스에서 정의된 메소드를 수정 또는 추가하는 기능(재정의)
class Cat(Animal):
    def sound(self):
        print("야옹") # 재정의

nabi = Cat()
nabi.sound()

class Person:
    def __init__(self, name, age):
        self.name = age
        self.age = age

class Student(Person):
    def __init__(self, name, age, major):
        super().__init__(name, age) # super()를 통해서 부모클래스의 생성자를 접근해서 name 속성을 설정
        self.major = major

std1 = Student("손흥민", 35, "스포츠댄스")
print(std1.name)    # 부모클래스로 부터 물려받은 속성
print(std1.major)   # 자식 고유의 속성

class Secret:
    def __init__(self):
        self.__password = "1234"

my_password = Secret()
print(my_password._Secret__password)

class Wallet:
    def __init__(self):
        self.__money = 1000 # 숨겨진 데이터

    def get_money(self):    # Getter : 값을 안전하게 전달하는 용도
        return self.__money

    def set_money(self, amount): # Setter : 검증 로직을 추가
        if amount >= 0:
            self.__money = amount
        else:
            print("잘못된 값입니다.")

    
wallet = Wallet()
wallet.set_money(2000)
print(wallet.get_money())
```
